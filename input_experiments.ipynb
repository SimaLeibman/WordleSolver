```python
import pandas as pd
#pd.set_option("display.max_colwidth", None)

from collections import Counter
```

```python
df_in = pd.read_csv('data/experiments.csv')
```

```python
print(df_in[:3])
```

```python
exp_df = df_in.copy()
exp_df['pt'] = df_in['patterns'].str.split('|')
```

```python
exp_df[:3]
```

```python
df_in[:3]
```

```python
steps_df = exp_df.groupby('ind')['pt'].agg(list).reset_index(name="ll")
```

```python
steps_df['onell'] = steps_df['ll'].apply(lambda x: [item for sublist in x for item in sublist])

steps_df['cnt'] = steps_df['onell'].apply(Counter)
```

```python
print(steps_df['cnt'].to_string(index=False))
```

```python
steps_df['cntstr'] = steps_df['cnt'].to_string(index=False)
```
