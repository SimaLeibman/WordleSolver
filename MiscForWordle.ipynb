```python
import numpy as np
from collections import Counter

pattern_vals = {'grey':0, 'yellow':1, 'green':2}
color_letters = "SZG" # in Russian :)


def grey() -> int: 
    return pattern_vals['grey']


def yellow() -> int:
    return pattern_vals['yellow']


def green() -> int:
    return pattern_vals['green']
    

# for an index (assumend to be 0,1,2) returns the Russian first letter for the color
def get_color_letter(ind: int) -> str:
    if ind in range(0,4):
        return color_letters[ind]
    else:
        return "X"
    

def decrement_counter(ct: Counter, key: str) -> None:
    if key in ct:     
        ct[key] -= 1
        if ct[key] == 0:
            del ct[key]
```

```python
def get_match_pattern(guess: str, answer: str) -> (np.ndarray, str, str):

    available_letter_counts = Counter(answer)
    
    result = np.full((5,), grey())

    for i, (c, a) in enumerate(zip(guess, answer)):
        if c == a:
            result[i] = green()
            decrement_counter(available_letter_counts, a)

    miss = np.where(result == grey())[0]
    for i in miss:
        curr = guess[i]
        #print(f"{i =}, {curr =}, {available_letter_counts}")
        if curr in available_letter_counts:
            result[i] = yellow()
            decrement_counter(available_letter_counts, curr)
    
    return result, ''.join(map(str,result)), ''.join(map(get_color_letter, result))
    
```

```python
def compare(guess: str, solution: str): #returns a tuple of 5 integers representing the pattern of a guess given a solution. 2 = green 1 = yellow 0 = gray
    assert len(guess) == len(solution) == 5

    result = [0]*5
    used = [False]*5

    # Greens first
    for i in range(5):
        if guess[i] == solution[i]:
            result[i] = 2
            used[i] = True

    # Yellows second
    for i in range(5):
        if result[i] == 2:
            continue
        for j in range(5):
            if not used[j] and guess[i] == solution[j]:
                result[i] = 1
                used[j] = True
                break


    return tuple(result), ''.join(map(get_color_letter, result))
```

```python
answer = 'ababa'
guess = 'babab'
r,sr,cl = get_match_pattern(guess, answer)
```

```python
tpl, cl = compare(guess, answer)
#print(tpl)
print(cl)
```
