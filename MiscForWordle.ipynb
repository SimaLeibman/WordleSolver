```python
import numpy as np
from collections import Counter

pattern_vals = {'grey':0, 'yellow':1, 'green':2, 'invalid':-1}
color_letters = "SZG" # in Russian :)


def grey() -> int: 
    return pattern_vals['grey']


def yellow() -> int:
    return pattern_vals['yellow']


def green() -> int:
    return pattern_vals['green']


def invalid() -> int:
    return pattern_vals['invalid']

# for an index (assumend to be 0,1,2) returns the Russian first letter for the color
def get_color_letter(ind: int) -> str:
    if ind in range(0,4):
        return color_letters[ind]
    else:
        return "X"
    

def decrement_counter(ct: Counter, key: str):
    if key in ct:     
        ct[key] -= 1
        if ct[key] == 0:
            del ct[key]
```

```python
def get_match_pattern(candidate: str, answer: str) -> (np.ndarray, str, str):

    available_letter_counts = Counter(answer)
    
    #result=5*[pattern_vals['invalid']]
    result = np.full((5,), invalid())

    for i, (c, a) in enumerate(zip(candidate, answer)):
        if c == a:
            result[i] = green()
            decrement_counter(available_letter_counts, a)

    for i in np.where(result == invalid())[0]:
        curr = candidate[i]
        if curr in available_letter_counts:
            result[i] = yellow()
            decrement_counter(available_letter_counts, a)
        else:
            result[i] = grey()
        
    return result, ''.join(map(str,result)), ''.join(map(get_color_letter, result))
    
```

```python
candidate = 'ababa'
answer = 'babab'
print(f"{candidate = }, {answer = }")
r,sr,cl = get_match_pattern(candidate, answer)
print(r)
print(sr)
print(cl)
```
